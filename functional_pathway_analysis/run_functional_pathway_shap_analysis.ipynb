```python
import sys
from pathlib import Path

import numpy as np
import pandas as pd
import shap
from sklearn.preprocessing import StandardScaler

sys.path.append("..")
from modeling import COVARIATES, build_lgbm, preprocess_fold

DATA_DIR = Path("/home/ec2-user/studies/Haochen_food/code_micro/data")
TRAIN_DIR = DATA_DIR / "train_pathways"
OUT_DIR = TRAIN_DIR / "shap"
SHAP_SAMPLE_N = 1000

model_input = pd.read_csv(TRAIN_DIR / "model_input_diet_cov_functional_pathways_baseline.csv", low_memory=False)
summary = pd.read_csv(TRAIN_DIR / "functional_pathway_summary_baseline.csv")
permutation = pd.read_csv(TRAIN_DIR / "permutation_test" / "functional_pathway_permutation_pvalues_fdr_baseline.csv")
targets = permutation.loc[permutation["perm_significant_fdr"], "pathway"].tolist()
features = COVARIATES + [c for c in model_input.columns if c not in ["participant_id", *COVARIATES, *summary["pathway"]]]
x, _ = preprocess_fold(model_input[features], model_input[features])
x_sample = x.sample(n=min(SHAP_SAMPLE_N, len(x)), random_state=1)
```

```python
mean_abs_shap = {}
for target in targets:
    y = model_input[target].to_numpy(dtype=float).reshape(-1, 1)
    model = build_lgbm().fit(x, StandardScaler().fit_transform(y).ravel())
    mean_abs_shap[target] = np.abs(shap.TreeExplainer(model).shap_values(x_sample)).mean(axis=0)
mean_abs_shap = pd.DataFrame(mean_abs_shap, index=features)
spearman = model_input[features + targets].corr(method="spearman").loc[features, targets]

OUT_DIR.mkdir(parents=True, exist_ok=True)
mean_abs_shap.to_csv(OUT_DIR / "functional_pathway_shap_mean_abs_wide_baseline.csv")
spearman.to_csv(OUT_DIR / "functional_pathway_shap_feature_target_spearman_baseline.csv")
(mean_abs_shap * np.sign(spearman)).to_csv(OUT_DIR / "functional_pathway_shap_directional_mean_abs_wide_baseline.csv")
print(mean_abs_shap.mean(axis=1).sort_values(ascending=False).head(20))
```
